````python
import pandas as pd

data = {
    "name": ["Patrick Mahomes", "Josh Allen", "Lamar Jackson",
             "Bijan Robinson", "Jahmyr Gibbs", "Saquon Barkley",
             "Justin Jefferson", "Ja'Marr Chase", "CeeDee Lamb",
             "Amon-Ra St. Brown"],

    "team": ["KC", "BUF", "BAL", "ATL", "DET", "PHI", "MIN", "CIN", "DAL", "DET"],

    "position": ["QB", "QB", "QB", "RB", "RB", "RB", "WR", "WR", "WR", "WR"],

    "age": [31, 30, 29, 24, 24, 29, 27, 26, 27, 27],

    "yards": [1250, 1400, 1100, 580, 504, 620, 540, 680, 490, 510],

    "touchdowns": [10, 12, 8, 5, 6, 7, 4, 8, 3, 5]
}

df = pd.DataFrame(data)

df
````

## Boolean Mask (Filter Mask)

A Boolean mask is used to identify rows in a DataFrame that meet a specific condition.

For example, say we want to access a particular value from a column in our DataFrame.

### Syntax

```python
df["column_name"] == "value_you_are_looking_for"
```

This returns a Pandas Series containing Boolean values (`True` or `False`).

- `True` — The row meets the condition.
- `False` — The row does not meet the condition.

### Example

```python
filt = df["position"] == "WR"
```

Expected output:

```text
0    False
1    False
2    False
3    False
4    False
5    False
6     True
7     True
8     True
9     True
Name: position, dtype: bool
```

This is known as a **Boolean mask**, or a **filter mask**.

The mask identifies which rows meet our condition, but does not yet filter the DataFrame itself.

````python
df["position"] == "WR"
````

````python
filt = (df["position"] == "WR")
````

````python
df[filt]
````

### Applying a Boolean Mask

There are two ways to apply a Boolean mask to a DataFrame:

```python
df[filt]
```

Or:

```python
df.loc[filt]
```

Both methods return the rows where the Boolean mask evaluates to `True`.

However, `.loc[]` also allows us to select specific columns alongside the filtered rows.

For example:

```python
df.loc[filt, "name"]
```

This returns only the names of the players who meet the filter condition.

````python
df.loc[filt, "name"]
# filt selects rows where the condition is True
# "name" selects the column to return
````

### AND (&) and OR (|) Operators

- **AND (&):** Retrieves rows where both conditions are true.
- **OR (|):** Retrieves rows where at least one condition is true.

For example, we want to retrieve all players whose
position is wide receiver (WR) and whose age is 27.

````python
#AND
filt1 = (df["position"] == "WR") & (df['age'] == 27)
df.loc[filt1]
````

````python
#OR
filt1 =  (df["position"] == "WR") | (df["age"] == 27)
df.loc[filt1]
````

### NOT (~) Operator

If we want to exclude all wide receivers (WR) who are
27 years old, we place the NOT (~) operator before
the filter variable to reverse the condition.

This retrieves all rows that do not match both conditions.

````python
#NOT "~"
filt1 =  (df["position"] == "WR") & (df["age"] == 27)
df.loc[~filt1]
````

### Filtering Players by Yardage

Suppose we want to filter players based on a certain number of yards.

````python
over_600 = df["yards"] > 600
df.loc[over_600]

#if we wanted specific columns we would put the columns after the filter
df.loc[over_600, ["yards","touchdowns", 'position']]
````

## Filtering Multiple Values Using `.isin()`

Say we're working with a large dataset and we want to filter for multiple specific positions within the DataFrame.

Instead of writing multiple conditions using the OR operator (`|`), we can use the `.isin()` method.

The `.isin()` method checks whether each value in a column exists within a given list and returns a Boolean mask containing `True` and `False` values.

In this example, we want to filter for players who are either quarterbacks (QB) or running backs (RB).

````python
specific_pos = ["QB", "RB"]
filt = df["position"].isin(specific_pos)
````

````python
df.loc[filt,["name","position", "yards"]]
````

The `.isin()` method is particularly useful when filtering for multiple values, as it makes the code cleaner and more readable.

The following two expressions produce the same Boolean mask:

```python
filt = df["position"].isin(["QB", "RB"])
```

```python
filt = (df["position"] == "QB") | (df["position"] == "RB")
```

Using `.isin()` is more convenient when filtering for a larger number of values within the same column.

## Filtering Using `.str.contains()`

Say we're working with a large dataset and want to find a specific word or keyword within a column.

We can use the `.str.contains()` method to check whether a string contains a particular word or sequence of characters.

This returns a Boolean mask containing `True` and `False` values.

In this example, we want to find players whose names contain the word "Allen".

### Additional Parameters

- `case=False` — Makes the search case-insensitive, meaning "allen" and "Allen" are treated the same.
- `na=False` — Treats missing values as `False`, excluding them from the filtered results.
- `NaN` — Stands for Not a Number and is commonly used to represent missing values within a dataset.

````python
filt = df["name"].str.contains("allen", case=False, na=False)
#this is saying for the column(name) in the dataset, does it contain the word "allen".
# Case=False means doesnt matter about upper/lower case
#NaN means Not a Number
#Na=False means remove those values from when showing the results

df.loc[filt,"name"]
````
